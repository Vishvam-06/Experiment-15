```python
#Dataset

import pandas as pd
import numpy as np

data = {
    'Product':['Laptop','Mobile', 'Tablet', 'Headphones', 'Smartwatch', 'Camera'],
    'Price': [55000,20000,30000,2000,8000,45000],
    'Units_Sold': [25,60,35,80,50,15],
    'Discount': [10,15,12,20,18,8]
}
df = pd.DataFrame(data)
print(df)
df
```

```python
#1. Min-Max Normalization
# Min-Max Normalization for Price

df['Price_MinMax'] = (df['Price'] - df['Price'].min()) / (df['Price'].max() - df['Price'].min())

print(df)
df

#if to print only specific column --> print(df[['Product','Price','Price_MinMax']])
```

```python
#2. Z-score Normalization
df['Units_Zscore'] = (df['Unit_Sold'] - df['Unit_Sold'].mean)
```

```python
#Normalization Multiple columns
cols = ['Price', 'Units_Sold','Discount']

df_norm = (df[cols] - df[cols].min()) / (df[cols].max() - df[cols].min())
df_norm
```

```python
#2.Z-score Normalization
df['Units_Zscore'] = (df['Units_Sold'] - df['Units_Sold'].mean()) / df['Units_Sold'].std()
print(df[['Product', 'Units_Sold', 'Units_Zscore']])
```

```python
#3. Decimal scaling

df['Price_Decimal'] = df['Price'] / 100000

print(df[['Product', 'Price', 'Price_Decimal']])
```

```python
import pandas as pd
import numpy as np

# Load the Amazon dataset
df = pd.read_csv('amazon_products_dataset_Expt-14.csv')
print(df.head())
```

```python
# Min-Max Normalization for Price
df['Price_MinMax'] = (df['Price'] - df['Price'].min()) / (df['Price'].max() - df['Price'].min())

print(df[['Product_Name', 'Price', 'Price_MinMax']].head())
```

```python
# Z-score Normalization for Units_Sold
df['Units_Zscore'] = (df['Units_Sold'] - df['Units_Sold'].mean()) / df['Units_Sold'].std()

print(df[['Product_Name', 'Units_Sold', 'Units_Zscore']].head())
```

```python
# Normalization for Multiple columns
cols = ['Price', 'Rating', 'Reviews', 'Units_Sold']

df_norm = (df[cols] - df[cols].min()) / (df[cols].max() - df[cols].min())
print(df_norm.head())
```

```python
# Decimal scaling for Reviews (Dividing by 10,000 as the reviews are mostly in thousands/hundreds)
df['Reviews_Decimal'] = df['Reviews'] / 10000

print(df[['Product_Name', 'Reviews', 'Reviews_Decimal']].head())
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

# Applying Label Encoding on Product_Name
df['Product_Name_Label'] = le.fit_transform(df['Product_Name'])

print(df[['Product_Name', 'Product_Name_Label']].head())
```

```python
# One-hot Encoding on Product_Name
# Note: Since all 50 products are unique, this will create 50 new dummy columns.
df_encoded = pd.get_dummies(df, columns=['Product_Name'])

print(df_encoded.head())
```

```python
# Dummy Encoding (Drop First) to prevent the dummy variable trap
df_dummy = pd.get_dummies(df, columns=['Product_Name'], drop_first=True)

print(df_dummy.head())
```
